## Python coding practice — 8 problems

```python
employees = [
    {"id": 1, "name": "Ali", "salary": 50000},
    {"id": 2, "name": "John", "salary": 60000},
    {"id": 3, "name": "Sara", "salary": 70000},
    {"id": 4, "name": "David", "salary": 45000}
]
```

Problem 1 — Find an employee by ID

Write a function that accepts an ID and returns the matching employee.

```python
def find_employee(employee_id):
    for emp in employees:
        if emp["id"] == employee_id:
            return emp
    return "employee not found"
print(find_employee(2))
```

Problem 2 — Filter employees by salary

Return employees whose salary is greater than or equal to 55000.

```python
result = []

for emp in employees:
    salary = emp["salary"]
    if salary >= 50000:
        result.append(emp)
print(result)
```

Problem 3 — Create a new employee

Write a function that accepts an employee name, ID and salary, then adds the employee to the list.

```python
def add_employee(employee_id, name, salary):
    new_employee = {
        "id":employee_id,
        "name":name,
        "salary": salary
    }
    employees.append(new_employee)

add_employee(5,"sam",60000)
print(employees)
```
