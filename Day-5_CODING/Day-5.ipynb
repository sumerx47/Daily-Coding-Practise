## Python coding practice — 8 problems

```python
employees = [
    {"id": 1, "name": "Ali", "salary": 50000},
    {"id": 2, "name": "John", "salary": 60000},
    {"id": 3, "name": "Sara", "salary": 70000},
    {"id": 4, "name": "David", "salary": 45000}
]
```

Problem 1 — Find an employee by ID

Write a function that accepts an ID and returns the matching employee.

```python
def find_employee(employee_id):
    for emp in employees:
        if emp["id"] == employee_id:
            return emp
    return "employee not found"
print(find_employee(2))
```

Problem 2 — Filter employees by salary

Return employees whose salary is greater than or equal to 55000.

```python
result = []

for emp in employees:
    salary = emp["salary"]
    if salary >= 50000:
        result.append(emp)
print(result)
```

Problem 3 — Create a new employee

Write a function that accepts an employee name, ID and salary, then adds the employee to the list.

```python
def add_employee(employee_id, name, salary):
    new_employee = {
        "id":employee_id,
        "name":name,
        "salary": salary
    }
    employees.append(new_employee)

add_employee(5,"sam",60000)
print(employees)
```

Problem 4 — Update employee salary

Update the salary of an employee using their ID.

```python
def update_salary(employee_id, new_salary):
    for emp in employees:
        if emp["id"] == employee_id:
            emp["salary"] = new_salary
            return emp
    return "employee not found"

print(update_salary(2,81000))
print(employees)
```

Problem 5 — Delete an employee

```python
def delete_employee(employee_id):
    for emp in employees:
        if emp["id"] == employee_id:
            employees.remove(emp)
            return "Employee deleted"

    return "Employee not found"

print(delete_employee(1))
print(employees)
```

Problem 6 — Validate employee input

Create a function that checks whether the name is present and salary is positive.

```python
def validate_employee(name,salary):
    if not name:
        return "Name is required"
    if salary <= 0:
        return "salary must be positive"

    return "valid employee"

print(validate_employee("Ali",50000))
```
