## Day 4 — PostgreSQL intensive 

```python
employees = [
    {"name": "Ali", "salary": 50000},
    {"name": "John", "salary": 60000},
    {"name": "Sara", "salary": 70000},
    {"name": "David", "salary": 45000}
]
```

Problem 1 — Filter employees by salary

Print employees whose salary is greater than 50000.

```python
result = []
for emp in employees:
    salary = emp["salary"]
    if salary > 50000:
        result.append(emp["name"])
print(result)
```

Problem 2 — Calculate average salary
Calculate the average salary of all employees without using a database.

```python
total = 0
for emp in employees:
    total += emp["salary"]

average = total/len(employees)
print(average)
```

Problem 3 — Group employees by department

```python
employees = [
    {"name": "Ali", "department": "IT"},
    {"name": "John", "department": "HR"},
    {"name": "Sara", "department": "IT"},
    {"name": "David", "department": "Finance"},
    {"name": "Ahmed", "department": "HR"}
]
```

```python
grouped = {}
for emp in employees:
    department = emp["department"]

    if department not in grouped:
        grouped[department] = []

    grouped[department].append(emp["name"])

print(grouped)
```

Problem 4 — Find duplicate salaries

```python
salaries = [50000, 60000, 50000, 70000, 60000, 45000]
```

```python
duplicates = []
for salary in salaries:
    if salaries.count(salary) > 1 and salary not in duplicates:
        duplicates.append(salary)

print(duplicates)
```

Problem 5 — Find the second-highest salary

```python
unique_salaries = sorted(set(salaries), reverse = True)
print(unique_salaries[1])
```

Problem 6 — Sort employees by salary
Sort employees from highest salary to lowest salary.

```python
employees = [
    {"name": "Ali", "salary": 50000},
    {"name": "John", "salary": 60000},
    {"name": "Sara", "salary": 70000}
]
```

```python
result = sorted(employees, key = lambda x:x["salary"] , reverse = True)
print(result)
```

Problem 7 — Count employees in each department

```python
employees = [
    {"name": "Ali", "department": "IT"},
    {"name": "John", "department": "HR"},
    {"name": "Sara", "department": "IT"},
    {"name": "David", "department": "Finance"},
    {"name": "Ahmed", "department": "HR"}
]
```

```python
count = {}

for emp in employees:
    department = emp["department"]

    if department not in count:
        count[department] = 0

    count[department] += 1
print(count)
```

Problem 8 — Create a simple employee search function
Write a function that accepts an employee name and returns the employee's details.

```python
employees = [
    {"id": 1, "name": "Ali", "salary": 50000},
    {"id": 2, "name": "John", "salary": 60000},
    {"id": 3, "name": "Sara", "salary": 70000}
]
```

```python
def find_employee(name):
    for emp in employees:
        if emp["name"].lower() == name.lower():
            return emp

    return "employee not found"

print(find_employee("Sara"))
print(find_employee("Saara"))
```
