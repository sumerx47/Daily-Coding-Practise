Problem 1 — Create a function with *args

Write a function that accepts multiple numbers and returns their total.

```python
def calculate_total(*args):
    total = 0
    for num in args:
        total += num
    return total

print(calculate_total(10,20,30))
```

Problem 2 — Use **kwargs

Create a function that accepts employee information.

```python
def employee_details(**kwargs):
    for key,value in kwargs.items():
        print(key,":",value)
employee_details(
    name="Ali",
    salary=50000,
    department= "IT"
)
```

Problem 3 — Create a class

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary

    def display(self):
        return {
            "name":self.name,
            "salary":self.salary
        }
employee = Employee("Ali",50000)
print(employee.display())
```

Problem 4 — Use inheritance

```python
class Employee:
    def __init__(self,name):
        self.name = name

class Developer(Employee):
    def __init__(self,name,language):
        super().__init__(name)
        self.language = language

developer = Developer("Sam","python")
print(developer.name)
print(developer.language)
```

Problem 5 — Create a custom exception

```python
class InvalidSalaryError(Exception):
    pass

def check_salary(salary):
    if salary <= 0:
        raise InvalidSalaryError("salary not found")

    return "valid salary"
    
print(check_salary(5000))  
```

Problem 6 — Create a decorator

```python
from functools import wraps

def login_required(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        print("Checking login")
        return function(*args, **kwargs)
    return wrapper


@login_required
def dashboard():
    return "Welcome to dashboard"

print(dashboard())
```

Problem 7 — Use a generator

Generate employee IDs from 1 to 5.

```python
def generate_ids():
    for number in range(1,11):
        yield number

for employee_id in generate_ids():
    print(employee_id)
```

Problem 8 — Handle exceptions

```python
def divide_numbers(a,b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        return "cannot divide by zero"

print(divide_numbers(2,5))
```
