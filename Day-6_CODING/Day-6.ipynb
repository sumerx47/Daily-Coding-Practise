Problem 1 — Create a function with *args

Write a function that accepts multiple numbers and returns their total.

```python
def calculate_total(*args):
    total = 0
    for num in args:
        total += num
    return total

print(calculate_total(10,20,30))
```

Problem 2 — Use **kwargs

Create a function that accepts employee information.

```python
def employee_details(**kwargs):
    for key,value in kwargs.items():
        print(key,":",value)
employee_details(
    name="Ali",
    salary=50000,
    department= "IT"
)
```

Problem 3 — Create a class

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary

    def display(self):
        return {
            "name":self.name,
            "salary":self.salary
        }
employee = Employee("Ali",50000)
print(employee.display())
```

Problem 4 — Use inheritance

```python
class Employee:
    def __init__(self,name):
        self.name = name

class Developer(Employee):
    def __init__(self,name,language):
        super().__init__(name)
        self.language = language

developer = Developer("Sam","python")
print(developer.name)
print(developer.language)
```
