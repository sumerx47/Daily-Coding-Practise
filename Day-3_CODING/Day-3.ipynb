Problem 1 — Handle multiple exceptions
Interview question: Write a function that divides two numbers and handles invalid input and division by zero.

```python
def divide_numbers(a,b):
    try:
        result = a/b
        return result

    except ZeroDivisionError:
        return "cannot divide by zero"

    except TypeError:
        return "invslid data type"

print(divide_numbers(10,2))
print(divide_numbers(10,0)) 
```

Problem 2 — Read a file and count words
Interview question: Read a text file and count how many times each word occurs.

```python
frequency = {}

with open("sample.txt","r") as file:
    for line in file:
        words = line.lower().split()

        for word in words:
            if word in frequency:
                frequency[word] += 1

            else:
                frequency[word] = 1
print(frequency)
```

Problem 3 — Create a custom exception
Interview question: Create a custom exception that raises an error when employee salary is less than 10,000.

```python
class InvalidSalaryError(Exception):
    pass

def check_salary(salary):
    if salary < 10000:
        raise InvalidSalaryError("salary must be at least 10000")
    return "valid salary"

try :
    print(check_salary(15000))
except InvalidSalaryError as e:
    print(e)
```

Problem 4 — Implement inheritance and method overriding
Interview question: Create an Employee class and a Developer class that inherits from Employee.

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary

    def display(self):
        print(self.name,self.salary)

class Developer(Employee):
    def __init__(self,name,salary, language):
        super().__init__(name, salary)
        self.language = language

    def display(self):
        print(
            self.name,
            self.salary,
            self.language
        )
        
dev = Developer("Ali",50000,"python")
dev.display()
```

Problem 5 — Create a custom iterator
Interview question: Create an iterator that returns numbers from 1 to 5.



```python
class MyNumber:
    def __init__(self):
        self.number = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.number <= 5:
            value = self.number
            self.number += 1
            return value
        else:
            raise StopIteration

number = MyNumber()
for num in number:
    print(num)
```

```python
def even_numbers():
    for num in range(1,21):
        if num % 2 == 0:
            yield num
for number in even_numbers():
    print(number)
```

Problem 7 — Create a decorator to calculate execution time
Interview question: Create a decorator that calculates how long a function takes to execute.

```python
import time
from functools import wraps

def calculate_time(func):
    @wraps(func) #preserves useful metadata of the original function
    def wrapper(*args,**kwargs): #allow the decorator to work with different function arguments
        start = time.time() # captures current time
        result = func(*args, **kwargs)
        end = time.time()
        print("execution time:" , end - start) #calculates elapsed time

        return result
    return wrapper

@calculate_time
def process_data():
    time.sleep(2) #pauses execution for two seconds.
    print("processing completed")

process_data()
```

Problem 8 — Process a large log file using a generator
Interview question: Suppose a log file contains thousands of lines. Find only the lines containing the word ERROR without loading the entire file into memory.

```python
def read_errors(filename):
    with open(filename,"r") as file:
        for line in file:
            if "ERROR" in line:
                yield line.strip()

for error in read_errors("server.log"):
    print(error)
```
