Problem 1 — Find the second-highest salary

```python
employees = [
    {"name": "Ali", "salary": 50000},
    {"name": "John", "salary": 70000},
    {"name": "Sara", "salary": 60000},
    {"name": "David", "salary": 70000},
    {"name": "Ahmed", "salary": 45000}
]
```

```python
salaries = []
for emp in employees:
    salaries.append(emp["salary"])

unique_salaries = set(salaries)

sorted_salaries = sorted(unique_salaries, reverse = True)

second_highest = sorted_salaries[1]

print(second_highest)
```

2. Group employees by department

```python
employees = [
    {"name": "Ali", "department": "IT"},
    {"name": "Sara", "department": "HR"},
    {"name": "John", "department": "IT"},
    {"name": "David", "department": "HR"},
    {"name": "Ahmed", "department": "Finance"}
]
```

```python
department_groups = {}

for employee in employees:

    department = employee["department"]
    name = employee["name"]

    if department not in department_groups:
        department_groups[department] = []

    department_groups[department].append(name)

print(department_groups)
```

3. Count word frequency

```python
sentence = "python fastapi python sql fastapi python"

words = sentence.split()

word_count = {}

for word in words:

    if word not in word_count:
        word_count[word] = 1
    else:
        word_count[word] += 1

print(word_count)
```

4. Filter and sort server records

```python
servers = [
    {"name": "web01", "cpu": 85, "memory": 60},
    {"name": "web02", "cpu": 95, "memory": 90},
    {"name": "db01", "cpu": 40, "memory": 88},
    {"name": "cache01", "cpu": 35, "memory": 45}
]
```

```python
filtered_servers = []

for server in servers:

    if server["cpu"] >= 80 or server["memory"] >= 80:
        filtered_servers.append(server)

sorted_servers = sorted(
    filtered_servers,
    key=lambda server: server["cpu"] + server["memory"],
    reverse=True
)

for server in sorted_servers:
    print(server)
```
