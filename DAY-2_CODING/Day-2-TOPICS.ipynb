1. Lists
A list stores multiple values in one variable.

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[0])
print(numbers[-1])
```

Method	  ||  purpose

append()	->  Add one element

extend()   ->	Add multiple elements

insert()  ->	Insert at a position

remove()  ->	Remove a particular value

pop()  ->	Remove and return an element

sort() ->	Sort original list

reverse()  ->	Reverse original list

count()  ->	Count occurrences

index()  ->	Find element position

copy()  ->	Create shallow copy

```python
numbers = [10, 20, 30]

numbers.append(40)
print(numbers)

numbers.remove(20)
print(numbers)
```

2. Tuples
A tuple is similar to a list, but it cannot be modified after creation.

```python
employee = ("Ali", 50000, "IT")

print(employee[0])
print(employee[1])
```

Difference:

List	  ||  Tuple

[]	   ||  ()

Mutable	    ||  Immutable

Can add elements    || 	Cannot add elements

Can remove elements	    ||  Cannot remove elements

Useful for changing data    || 	Useful for fixed data

```python
employee = ("Ali", 50000, "IT")

name, salary, department = employee

print(name)
print(salary)
print(department)
```

3. Dictionaries
A dictionary stores data in key-value pairs.

```python
employee = {
    "name": "Ali",
    "salary": 50000,
    "department": "IT"
}

print(employee["name"])
print(employee["salary"])
```

Important methods:

Method	|| Purpose

keys() ||	Returns all keys

values() ||	Returns all values

items()	 || Returns key-value pairs

get() ||	Safely retrieves a value

update()	 || Updates dictionary

pop() ||	Removes a key

setdefault()	 || Gets key or inserts default

copy() ||	Creates shallow copy

```python
employee = {
    "name": "Ali", 
    "salary": 50000
    }

employee["salary"] = 60000
employee["department"] = "IT"

print(employee)
```

4. Sets
A set stores unique values. Duplicate elements are automatically removed.

```python
numbers = [10, 20, 20, 30, 30, 40]

unique = set(numbers)

print(unique)
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))
```

Set operation meanings
- Union → All unique elements from both sets.
- Intersection → Common elements.
- Difference → Elements in the first set but not the second.
- Symmetric difference → Elements not common to both.

5. Functions
A function is a reusable block of code that performs a particular task.

```python
def add(a, b):
    result = a + b
    return result

answer = add(10, 20)

print(answer)
```

Parameters vs arguments

```python
def greet(name):  # name = parameter
    print("Hello", name)

greet("Ali")      # Ali = argument
```

6. Lambda functions and comprehensions
Lambda: A small anonymous function written in one expression.

```python
# Normal function:
def square(x):
    return x * x

square(5)
```

```python
# Lambda equivalent:
square = lambda x: x * x

print(square(5))
```

List comprehension: A shorter way to create a list.

Normal method:

```python
numbers = [1, 2, 3, 4, 5]

result = []

for n in numbers:
    if n > 2:
        result.append(n)

print(result)
```

```python
''' List comprehension:

Remember this pattern:
[expression for item in iterable if condition] 
'''
numbers = [1, 2, 3, 4, 5]

result = [n for n in numbers if n > 2]

print(result)
```
