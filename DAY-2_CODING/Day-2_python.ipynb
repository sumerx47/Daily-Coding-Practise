Problem 1 — Find common elements between two lists
Interview question: Given two lists, find the elements present in both lists without duplicates.

```python
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]
```

```python
common = list(set(list1) & set(list2))
print(common)
```

```python
common = []

for num in list1:
    if num in list2 and num not in common:
        common.append(num)

print(common)
```

Problem 2 — Remove duplicates while preserving order

```python
numbers = [10, 20, 10, 30, 20, 40, 50, 30]
```

```python
result = []

for num in numbers:
    if num not in result:
        result.append(num)

print(result)
```

Problem 3 — Count frequency of each word
Input:

```python
words = ["python", "java", "python", "sql", "java", "python"]
```

```python
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1
print(frequency)
```

Problem 4 — Group employees by department
Input: Our employee dataset.


```python
employees = [
    {"name": "Ali", "department": "IT", "salary": 50000},
    {"name": "John", "department": "HR", "salary": 60000},
    {"name": "Sara", "department": "IT", "salary": 70000},
    {"name": "David", "department": "Finance", "salary": 45000},
    {"name": "Ahmed", "department": "HR", "salary": 60000},
    {"name": "Zara", "department": "IT", "salary": 80000}
]
```

```python
grouped = {}

for emp in employees:
    department = emp['department']

    if department not in grouped:
        grouped[department] = []

    grouped[department].append(emp['name'])

print(grouped)
```
