Q1. Find the student with the highest marks.

```python
students = {
    "Alice": 85,
    "Bob": 72,
    "Charlie": 90,
    "David": 72,
    "Eva": 85,
    "Frank": 95
}
```

```python
topper = max(students,key=students.get)
print(topper,students[topper])
```

Q2. Count how many students scored above 80.

```python
count = sum(1 for marks in students.values() if marks> 80)
print(count)
```

```python
count = 0
for marks in students.values():
    if marks > 80:
        count += 1
print(count)
```

Q3. Find students who scored at least 85.

```python
result = {
    name: marks 
    for name, marks in students.items()
    if marks >= 85
    }
print(result)
```

```python
result = {}
for name,marks in students.items():
    if marks >= 85:
        result[name] = marks

print(result)
```

Q4. Calculate average marks.

```python
average = sum(students.values())/len(students)
print(average)
```
