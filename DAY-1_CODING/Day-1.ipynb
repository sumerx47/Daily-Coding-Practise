Q1. Find the student with the highest marks.

```python
students = {
    "Alice": 85,
    "Bob": 72,
    "Charlie": 90,
    "David": 72,
    "Eva": 85,
    "Frank": 95
}
```

```python
topper = max(students,key=students.get)
print(topper,students[topper])
```

Q2. Count how many students scored above 80.

```python
count = sum(1 for marks in students.values() if marks> 80)
print(count)
```

```python
count = 0
for marks in students.values():
    if marks > 80:
        count += 1
print(count)
```

Q3. Find students who scored at least 85.

```python
result = {
    name: marks 
    for name, marks in students.items()
    if marks >= 85
    }
print(result)
```

```python
result = {}
for name,marks in students.items():
    if marks >= 85:
        result[name] = marks

print(result)
```

Q4. Calculate average marks.

```python
average = sum(students.values())/len(students)
print(average)
```

Q5. Sort students by marks in descending order.

```python
result = sorted(students.items(), key = lambda x:x[1], reverse =True)
print(result)
```

Q6. Group students by marks.

```python
grouped = {}

for name,marks in students.items():
    grouped.setdefault(marks,[]).append(name)
print(grouped)
```

```python
grouped = {}

for name,marks in students.items():
    if marks not in grouped:
        grouped[marks] = []

    grouped[marks].append(name)
print(grouped)
```

Q7. Find the second-highest unique mark.

```python
unique_marks = sorted(set(students.values()), reverse = True)
print(unique_marks[1])
print(unique_marks)
```

Q8. Find duplicate marks and their frequency.

```python
from collections import Counter

frequency = Counter(students.values())

duplicates = {
    mark:count
    for mark, count in frequency.items()
    if count > 1
}

print(duplicates)
```

```python
frequency = {}

for marks in students.values():
    if marks in frequency:
        frequency[marks] += 1
    else:
        frequency[marks] = 1

print(frequency)

duplicates = {}

for mark, count in frequency.items():
    if count > 1:
        duplicates[mark] = count

print(duplicates)
```
